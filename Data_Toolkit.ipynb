```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample DataFrame
data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 25000, 30000, 35000, 40000],
    "Experience": [1, 3, 5, 7, 9]
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Visualize using heatmap
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python

import plotly.express as px
import pandas as pd

# Create sample data
data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)

# Create bar plot
fig = px.bar(
    df,
    x="Name",
    y="Marks",
    title="Student Marks",
    labels={"Name": "Student Name", "Marks": "Marks"}
)

# Display the plot
fig.show()
```

```python

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]
y3 = [2, 3, 5, 8, 12]

# Plot multiple lines
plt.plot(x, y1, label="Line 1", marker="o")
plt.plot(x, y2, label="Line 2", marker="s")
plt.plot(x, y3, label="Line 3", marker="^")

# Add title, labels, and legend
plt.title("Multiple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()

# Display the plot
plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80,
         82, 85, 88, 90, 92, 95, 60, 70, 75, 80]

# Create histogram
sns.histplot(data=marks, bins=5, kde=True)

# Add title and labels
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Display plot
plt.show()
```

```python

import numpy as np

# Create two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Perform matrix multiplication
result = np.matmul(A, B)

print("Matrix A:")
print(A)

print("Matrix B:")
print(B)

print("Matrix Multiplication:")
print(result)
```

```python

import plotly.express as px
import pandas as pd

# Create sample data
data = {
    "X": [1, 2, 3, 4, 5],
    "Y": [5, 3, 4, 2, 1],
    "Z": [2, 4, 1, 5, 3]
}

df = pd.DataFrame(data)

# Create 3D scatter plot
fig = px.scatter_3d(
    df,
    x="X",
    y="Y",
    z="Z",
    title="3D Scatter Plot"
)

# Display the plot
fig.show()
```
